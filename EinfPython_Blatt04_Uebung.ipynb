Die spezielle Syntax `*args` für Positionsparameter und `**kwargs` für Schlüsselwortparameter in Funktionsdefinitionen wird in Python dazu genutzt, eine beliebige Anzahl von Argumenten zu übergeben. Die Bezeichner `args` und `kwargs` können durch andere Bezeichner ersetzt werden.  
Jeweils referenzieren `args` ein Tupel und `kwargs` ein Dictionary.

### a) Variable Anzahl an Positionsparametern (`*`-Parameter)
Schreiben Sie eine Funktion `addiere` für Zahlen, die beliebig viele Positionsparameter erwartet, indem Sie die oben erklärte Syntax benutzen, und die Summe aller übergebenen Argumente zurückgibt.

### b) Unpacking Operator (`*`-Argument)
Will man eine Liste von Positionsargumenten (und nicht die Liste selbst) an eine Funktion weitergeben, so wird der *Unpacking-Operator* `*` benötigt. Mehr Informationen finden Sie [hier](https://docs.python.org/3/tutorial/controlflow.html#unpacking-argument-lists).

Verwenden Sie die Funktion `addiere` aus Aufgabenteil a), um die Summe der Zahlen in der Liste `lst` zu bestimmen, indem Sie den Unpacking Operator verwenden.

````python
lst = [1, -2, 3, 7, 5, 2]
````

Betrachten Sie untenstehenden Code. Was sagt dieser über das Verwenden von Unpacking bei einer festen Parameteranzahl aus? 

````python
def g(a, b, c, d):
    return a / b * (c+d)

lst = [5, 1, 3, 5]

print("g(5, 1, 3, 5):", g(5, 1, 3, 5))
print("g(*[5, 1, 3, 5]):", g(*lst))
print("g(5, *[1, 3, 5]):", g(5, *lst[1:]))

lst.append(10)

print(g(*lst))
````

Zwei Faustregeln für gute Programmierpraxis in Python lauten:  
- Der Datenaustausch mit Funktionen sollte ausschließlich über Parameter und Rückgabewerte erfolgen.  
- Explizit ist besser als implizit.   

Globale Referenzen widersprechen direkt dem ersten Punkt, denn durch globale Referenzen ist bei Ausführung einer Funktion nicht ersichtlich, woher die Instanz bezogen wird. Python verlangt deshalb explizit das keyword `global`, um daran zu erinnern, dass eine globale Referenz verwendet wird.

Programme, die viele globale Referenzen benutzen sind in der Regel schlechter testbar, denn der Rückgabewert einer Funktion hängt in diesem Fall sowohl von den Parametern sowie vom Zustand des globalen Namenraumes ab.

---

*Nice to Know:  
Mithilfe der parameterlosen eingebauten Funktionen `globals` bzw. `locals` können Sie sich den aktuellen globalen bzw. lokalen Namensraum als Dictionary zurückgeben lassen.*

*Bermerkung:  
Im Jupyter-Notebook werden globale Referenzen beibehalten (die Referenzen sind über alle ausgeführten Zellen hinweg "sichtbar").
Deshalb liefert `globals()` ebenfalls alle zuvor definierten Referenzen.*

````python
# Beispiel zur zweiten Teilaufgabe:
a = "Hallo"
def f():
    print("globals()['a'] =", globals()['a'])
    print("a in locals() =", ("a" in locals()), "\n")

    a = "Welt"
    print("after: a = 'Welt':")
    print("globals()['a'] =", globals()['a'])
    print("a in locals() =", ("a" in locals()) )
    print("locals()['a'] =", locals()['a'], "\n")

    print(a)
f()
print(a)
````

### d) Rekursive Funktionen
Eine Funktion, die sich selbst aufruft, wird *rekursiv* genannt.  
Die folgende rekursive Funktion `fak` berechnet die Fakultät einer natürlichen Zahl.

````python
def fak(n):
    if n > 1:
        return n * fak(n-1)
    return 1
````
