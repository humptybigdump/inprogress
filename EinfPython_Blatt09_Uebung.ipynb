## Aufgabe 1 (Iteratoren und Generatoren - Grundlagen)

Ein iterierbares Objekt `iterable` ist eine Instanz eines Datentyps, für welche die Methode `__iter__` implementiert ist. Diese magische Methode wird durch `iter(iterable)` aufgerufen und liefert einen Iterator des iterierbaren Objekts `iterable`. Mit diesem kann mithilfe der Funktion `next` ein Objekt nach dem Anderen durchlaufen und zur Verfügung gestellt werden werden.

Die Funktion `next` steht nur für Iteratoren zur Verfügung, kann also nicht direkt  z.B. auf eine Liste angewendet werden.

Iteratoren sind spezielle iterierbare Objekte, da sie wiederum die magische Methode `__iter__` implementieren, jedoch einen speziellen Rückgabewert aufweisen (siehe unten). Zusätzlich ist die Methode `__next__` implementiert.

Generatoren sind spezielle Iteratoren, welche jeweils beim `next`-Aufruf das nächste Objekt generieren ohne alle Objekte als Ganzes zu speichern. Generatoren können durch Klassen, Generatorfunktionen und Generator-Expressions realisiert werden.
