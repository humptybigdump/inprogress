#### Bonus Task to part c: Fully automate the identification of the best amount and combination of warehouses



```python
from itertools import combinations
# We try every set of warehouses, assign demands, and shift load with minimal penalty.

num_warehouses = len(distances)
num_customers = len(demand)
capacity = 100

#track the best solution across all subsets ---
best_cost = float('inf')
best_wh_ids = None

# Try selecting r warehouses, where r ranges from 1 up to the total number
for r in range(1, num_warehouses + 1):
    # For each r, generate all unique subsets of warehouse indices of size r
    for subset in combinations(range(num_warehouses), r):

        # Convert to 1-based IDs for printing
        wh_ids = [w + 1 for w in subset]
        print("\nEvaluating warehouses:", wh_ids)

        # 1) Build preference lists for each customer
        preferences = []
        for cust in range(num_customers):
            # Sort this subset by distance for customer `cust`
            sorted_wh = sorted(subset, key=lambda w: distances[w][cust])
            preferences.append(sorted_wh)
            print(f" Customer {cust+1} prefers warehouses:", [w+1 for w in sorted_wh])

        # 2) Initialize flows and remaining capacities
        flows = [[0]*num_customers for _ in range(num_warehouses)]
        remaining_capacity = [capacity]*num_warehouses

        # Assign each customer's full demand to its cheapest warehouse
        for cust in range(num_customers):
            best_wh = preferences[cust][0]
            flows[best_wh][cust] = demand[cust]
            remaining_capacity[best_wh] -= demand[cust]
        print(" Initial capacities after assignment:", {w+1: remaining_capacity[w] for w in subset})

        # 3) Fix overloads by shifting the minimal-penalty amount
        feasible = True
        while True:
            overloaded = [w for w in subset if remaining_capacity[w] < 0]
            if not overloaded:
                break  # no more overloads

            print(" Overloaded:", [w+1 for w in overloaded])
            # Collect possible shifts: (penalty, from_wh, cust, to_wh)
            shifts = []
            for wh in overloaded:
                excess = -remaining_capacity[wh]
                for cust in range(num_customers):
                    alloc = flows[wh][cust]
                    if alloc == 0:
                        continue
                    pref_list = preferences[cust]
                    idx = pref_list.index(wh)
                    if idx+1 < len(pref_list):
                        next_wh = pref_list[idx+1]
                        pen = (distances[next_wh][cust] - distances[wh][cust]) * unit_shipping_cost
                        shifts.append((pen, wh, cust, next_wh))

            if not shifts:
                print(" No valid shifts → infeasible subset")
                feasible = False
                break

            shifts.sort(key=lambda x: x[0])
            penalty, from_wh, cust, to_wh = shifts[0]
            shift_amt = min(flows[from_wh][cust], -remaining_capacity[from_wh])
            print(f"  Shifting {shift_amt} units of customer {cust+1} "
                  f"from WH{from_wh+1} to WH{to_wh+1} (penalty/unit €{penalty:.2f})")

            # Apply the shift
            flows[from_wh][cust] -= shift_amt
            flows[to_wh][cust] += shift_amt
            remaining_capacity[from_wh] += shift_amt
            remaining_capacity[to_wh] -= shift_amt
            print("  Updated capacities:", {w+1: remaining_capacity[w] for w in subset})

        if not feasible:
            continue

        # 4) Compute total cost: fixed + shipping
        fixed_total = r * fixed_cost
        shipping_total = 0
        for wh in subset:
            for cust in range(num_customers):
                shipping_total += flows[wh][cust] * distances[wh][cust] * unit_shipping_cost

        total = fixed_total + shipping_total
        print(f" Total cost for warehouses {wh_ids}: €{total:.2f}")

        # update best if this is the cheapest so far
        if total < best_cost:
            best_cost = total
            best_wh_ids = wh_ids.copy()

# After checking all subsets, print the overall best
print("\n" + "="*50)
print("Optimal solution found:")
print(" Warehouses to open:", best_wh_ids)
print(f" Total monthly cost: €{best_cost:.2f}")

```
