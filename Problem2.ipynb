Functions in Python are initialized by command ``def`` followed by the name of the function, function arguments in parantheses and a colon (:), e.g.: ``def function_name(arg):`` Additionally, the command ``return`` indicates the objects that should be returned to the calling code.

**Example:**

The code below defines a function called ``myFunction`` that takes one argument, prints the argument, and returns a variable ``x``. Once the function is defined, it can be called elsewhere in the code.

```python
# define myFunction
def myFunction(arg0):
    # print the input argument arg0
    print(arg0)

    # create a variable x
    x = 10

    # return the value of x
    return x

# use myFunction in the code
value = 15
y = myFunction(value)
print(y)
```

**Example 1:**

You can access individual elements of an array by specifying their indices using square brackets. To extract entire sequences, you can use slicing with the colon (:) operator.

```python
array = np.array([1, 2, 3, 4, 5, 6, 7])

print(array[0]) # access entries using the index
print(array[3])

print(array[:]) # access multiple entries using slicing
print(array[2:5])
```
