g) The syntax to call the routine is then ``spo.fsolve(in1, in2, in3)``. In order, we have to pass the following inputs to the function ``fsolve``: ``in1``: The function, ``in2``: An initial guess for the root, ``in3``: Additional parameters of the defined function, which can be done by using syntax ``args=()``. Store the result in the objects ``root, info, ier`` and ``msg``.

**Example:**

```python
import scipy.optimize as spo

# define a function (a and b are parameters)
def g(x, a, b): 
    return a * x + b

a = 2 # set parameters
b = 5

x0 = -2.0 # initial guess

root, info, ier, msg = spo.fsolve(g,  x0, args=(a, b), full_output=True)
print(root)
```
