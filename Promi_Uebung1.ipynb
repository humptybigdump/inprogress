```python
"""
Import the necessary libraries 
pm4py is a process mining library
pandas is a data manipulation library
"""

import pm4py
import pandas as pd
```

```python
def read_process_doku(file_path):
    """
    Reads the process documentation from a CSV file.

    Parameters:
    file_path (str): The path to the CSV file.

    Returns:
    pandas.DataFrame: The processed documentation data.
    """
    process_doku = pd.read_csv(file_path, sep=',')
    return process_doku

process_doku = read_process_doku('KKR_Prozessdokumentation.csv')
```

```python
# Display the first 5 rows of the process documentation
process_doku.head(5)
```

```python
# Convert the 'Zeitstempel' column to datetime and the 'Fall_ID' column to string
process_doku['Zeitstempel'] = pd.to_datetime(process_doku['Zeitstempel'])
process_doku['Fall_ID'] = process_doku['Fall_ID'].astype(str)
```

```python
# Renames the columns of the 'process_doku' dataframe.
process_doku.rename(columns={'Zeitstempel': 'time:timestamp', 'Fall_ID': 'case:concept:name', 'Aktivitaet': 'concept:name'}, inplace=True)
```

```python
def get_num_cases(process_doku):
    """
    Calculates the number of unique cases in the given process documentation.

    Args:
        process_doku (DataFrame): The process documentation DataFrame.

    Returns:
        int: The number of unique cases.
    """
    num_cases = process_doku['case:concept:name'].nunique()
    return num_cases

# Get the number of cases in the process documentation
num_cases = get_num_cases(process_doku)
print("Number of cases:", num_cases)
```

```python
# Get the start and end activities of the process
start_activities = pm4py.get_start_activities(process_doku)
end_activities = pm4py.get_end_activities(process_doku)

# Print the start activities
print("\033[1mStart activities | Anzahl Fälle:\033[0m")
print("__________________________________")
for activity in start_activities.keys():
    print(activity, "|", start_activities.get(activity))

# Print a newline    
print("\n")

# Print the end activities
print("\033[1mEnd activities | Anzahl Fälle:\033[0m")
print("__________________________________")
for activity in end_activities.keys():
    print(activity, "|", end_activities.get(activity))
```

```python
# Get the event attributes of the process documentation
event_attributes = pm4py.stats.get_event_attributes(process_doku)

# Print the event attributes
for attribute in event_attributes:
      print(attribute)
```

```python
# Get the case durations from the process documentation
case_durations = pm4py.stats.get_all_case_durations(process_doku)
# Calculate the duration in minutes
case_durations = [round((duration/60)) for duration in case_durations]

# Print the case durations per case
for i,duration in enumerate(case_durations):
    print(f"{i+1}. Fall:", duration, 'min')
```
