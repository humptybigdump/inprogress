```python
# Let's say, we want to check the sequence for the number of adenines
seq = 'ACTGGAGTCAGG'

if not len(seq):
    print('The DNA sequence is empty')
    print('Please enter a valid DNA sequence')
else:
    count_a = seq.count('A')
    print('Adenine count: ', count_a)
    print('+++ Sequence analysed +++')
```

```python
seq = input('Please enter a DNA sequence: ')
seq = seq.replace(" ", "")
seq = seq.upper()

if not len(seq):
    print('The DNA sequence is empty')
    print('Please enter a valid DNA sequence')
else:
    freq_a = seq.count('A') / len(seq)
    print('Adenine frequency: ', freq_a)
    print('+++ Sequence analysed +++')

# potential problems in this code?
# How can we eliminate potential problems before they occur?
```

```python
temperature = float( input("Please enter a number") )

if temperature > 25:
    print('It is hot.')
elif temperature > 15:
    print('It is warm.')
else:
    print('It is cold.')
```

```python
# wthout the use of elif, we "gain" one more identation level that should be avoided
temperature = 15

if temperature > 25:
    print('It is hot.')
else:
    if temperature > 15:
        print('It is warm.')
    else:
        print('It is cold.')
```
