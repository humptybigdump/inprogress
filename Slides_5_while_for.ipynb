```python
# using range()
# range() can take 1, 2, or 3 input values
# 1 input value
for a in range(11):
    print(a, a ** 2)
```

```python
# 2 input values
for a in range(1,101):
    print(a, a**2)
```

```python
# 3 input values
for a in range(1,9,2):
    print(a*2)
```

```python
lst = ['a', 'b', 'c', 'd', 'e']

# without enumerate
counter = 0
for item in lst:
    print(counter, item)
    counter += 1

print()

# with enumerate
for i, item in enumerate(lst):
    print(i, item)

print()

# use range and indices
for i in range( len(lst) ):
    print(i, lst[i])
```

```python
# List of DNA sequences
sequences = [
    "ATCGTA",
    "GGGCCC",
    "TTAACG",
    "GTCAGC",
    "AAATTT"
]

# Go through each sequence
for seq in sequences:
    # Count number of guanines in each nucleotide
    G_count = 0
    for nucleotide in seq:
        if nucleotide == 'G':
            G_count += 1
    print(seq, G_count)
```

```python
# or a little shorter
for seq in sequences:
    print(seq, seq.count('G'))
```
