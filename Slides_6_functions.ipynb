```python
def even(a):
    if a%2 == 0:
        return "Even"
    else:
        return "Uneven"

lst1 = [1,3,6,9,8,100,101,100005]
for num in lst1:
    print(even(num))
```

```python
def even_number(a):
    if a%2 == 0:
        return "even"
    else:
        return "uneven"

lst = [1,3,6,9,8,100,101,100005,"seven"]

for num in lst:
    print(even_number(num))
```

```python
def modulo(m,n):
    quot = m // n
    remd = m % n

    return quot, remd

a, b = modulo(11,3)
print(a,b)
```

```python
# first build without function
DNA_sequence = "ATCGTAGC"
def complement_strand(seq):
    complement = {'A': 'T', 'T': 'A', 'C': 'G', 'G': 'C'}
    new_strand = ""
    for base in seq:
        new_strand += complement[base]
        print(base, "-->", complement[base], ": ", new_strand)
    return new_strand

print(complement_strand(DNA_sequence))
# Output: TAGCATCG
```

```python
def is_valid_DNA(seq):
    for nucleotide in seq:
        if nucleotide not in "ATCG":
            return False
    return True

print(is_valid_DNA("ATCGTAGC"))   # True
print(is_valid_DNA("ATBGTAGC"))   # False
```

```python
def process_sequence(seq):
    # 1. Transcribe DNA to RNA
    rna = seq.replace('T', 'U')
    # 2. Get length
    length = len(rna)
    # 3. Check if starts with "AUG"
    starts_with_aug = rna.startswith("AUG")
    return rna, length, starts_with_aug

dna1 = "ATGGCTA"
dna2 = "TTAACG"
dna3 = "GGGATG"

for dna in [dna1, dna2, dna3]:
    rna, length, starts_with_aug = process_sequence(dna)
    print(f"RNA: {rna}, Length: {length}, Starts with AUG: {starts_with_aug}")
```
