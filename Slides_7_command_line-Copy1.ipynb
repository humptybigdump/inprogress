```python
with open("P01308.fasta", "r") as f:
    # Function readline(consumes a saingle line
    data = f.readline()
    print(data)
```

```python
with open("P01308.fasta", "r") as f:
    line = f.readline()
    while line != '':
        print(line)
        line = f.readline()
```

```python
with open("P01308.fasta", "r") as f:
    line = f.readline()
    while line != '':
        print(line.strip())
        line = f.readline()
```

```python
# A script to copy a file to a another file
# We use a file for reading and for writing

lines = []

# Read from first file and store lines in a list
with open('P01308.fasta', 'r') as f:
    line = f.readline().strip()
    while line != '':
        lines.append(line)
        line = f.readline().strip()

# Write lines from the list into another file
with open('P01308-copy.fasta', 'w') as f:
    for line in lines:
        f.write(line + '\n') # Method write() writes a given string to the outfile
```

```python
# let's iterate over a folder of files

import os
path_to_files = "fasta_files/"
fasta_files = os.listdir(path_to_files)
for fasta in fasta_files:
    with open(path_to_files + fasta, 'r') as f:
        header = f.readline().strip()
        print(header)
```
