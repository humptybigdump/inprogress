```python
df = pd.read_csv("water_pollution_disease.csv")  # 2. Load a CSV file
df = df[["pH Level", "Bacteria Count (CFU/mL)", "GDP per Capita (USD)", "Urbanization Rate (%)"]].head(150)
```

```python
df["pH Level"].plot()
```

```python
df.plot(kind="bar")
```

```python
df["Bacteria Count (CFU/mL)"].plot(kind="bar")  # Different plot types: bar, barh, hist, pie, box
```
