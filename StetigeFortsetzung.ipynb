## Stetige Fortsetzung (Ergänzung zu den Beipielen im Skript):

#### a) Funktion $f:\mathbb R_+\to\mathbb R, \quad x\mapsto \exp(-1/x)$
* Die Funktion $f$ ist als Komposition stetiger Funktionen stetig. 
* Die Zahl $0$ ist Häufungspunkt der Definitionsmenge $\mathbb R_+$. 
* Lässt sich die Funktion in $x=0$ stetig fortsetzen? Der Graph legt dies nahe:

```python
fig, ax = plt.subplots()

# Parameter für die Darstellung der Achsen
ax.tick_params('x',labelsize=15)
ax.tick_params('y',labelsize=15)
ax.set_xlabel(r"$x$", fontsize=20) 
ax.set_ylabel(r"$\exp(-1/x)$", fontsize=20)

# Gitter auf dem Intervall [0.001, 5] mit 1000 Gitterpunkten
x = np.linspace(0.001, 5, 1000) 
# Plot der Funktion auf dem Gitter
ax.plot(x, np.exp(-1/x), color="k") 

plt.show()
```

**Mathematische Begründung:**
* Es gilt: $-1/x$ ist für $x\to 0^+$ uneigentlich konvergent gegen $-\infty$. 
* Daher folgt $\displaystyle\lim_{x\to 0} \exp(-1/x)=0$, da $\displaystyle\lim_{x\to-\infty} \exp(x)=0$ gilt.

#### b) Funktion $f:\mathbb R\setminus\{0\}\to\mathbb R, \quad x\mapsto \exp(-1/x)$
* Die Funktion $f$ ist als Komposition stetiger Funktionen stetig. 
* Die Zahl $0$ ist Häufungspunkt der Definitionsmenge $\mathbb R\setminus\{0\}$. 
* Lässt sich die Funktion in $x=0$ stetig fortsetzen? Nein. Begründen Sie anhand des Graphen.

```python
fig, ax = plt.subplots()

# Parameter für die Darstellung der Achsen
ax.tick_params('x',labelsize=15)
ax.tick_params('y',labelsize=15)
ax.set_xlabel(r"$x$", fontsize=20) 
ax.set_ylabel(r"$\exp(-1/x)$", fontsize=20)

# Gitter auf dem Intervall [0.001, 5] mit 1000 Gitterpunkten
x = np.linspace(0.001, 5, 1000) 
# Plot der Funktion auf dem Gitter
ax.plot(x, np.exp(-1/x), color="k") 
# Gitter auf dem Intervall [-5, -0.5] mit 1000 Gitterpunkten
x = np.linspace(-5, -0.5, 1000)
# Plot der Funktion auf dem Gitter
ax.plot(x, np.exp(-1/x), color="k")

plt.show()
```

#### c) Funktion $f:\mathbb R_+\to\mathbb R, \quad x\mapsto \sin(1/x)$
* Die Funktion $f$ ist als Komposition stetiger Funktionen stetig. 
* Die Zahl $0$ ist Häufungspunkt der Definitionsmenge $\mathbb R_+$. 
* Lässt sich die Funktion in $x=0$ stetig fortsetzen? Nein, für $x\to 0^+$ kann jeder Wert in $[-1,1]$ angenommen werden (betrachte dazu die Folge $x = \frac{1}{\varphi + 2\pi n}$):

```python
fig, ax = plt.subplots()

# Parameter für die Darstellung der Achsen
ax.tick_params('x',labelsize=15)
ax.tick_params('y',labelsize=15)
ax.set_xlabel(r"$x$", fontsize=20) 
ax.set_ylabel(r"$\sin(1/x)$", fontsize=20)

# Gitter auf dem Intervall [0.001, 0.5] mit 3000 Gitterpunkten
x = np.linspace(0.001, 0.5, 3000) 
# Plot der Funktion auf dem Gitter
ax.plot(x, np.sin(1/x), color="k") 

plt.show()
```

#### d) Funktion $f:\mathbb R_+\to\mathbb R, \quad x\mapsto x\sin(1/x)$
* Die Funktion $f$ ist als Komposition stetiger Funktionen stetig. 
* Die Zahl $0$ ist Häufungspunkt der Definitionsmenge $\mathbb R_+$. 
* Lässt sich die Funktion in $x=0$ stetig fortsetzen? 

```python
fig, ax = plt.subplots()

# Parameter für die Darstellung der Achsen
ax.tick_params('x',labelsize=15)
ax.tick_params('y',labelsize=15)
ax.set_xlabel(r"$x$", fontsize=20) 
ax.set_ylabel(r"$x\sin(1/x)$", fontsize=20)

# Gitter auf dem Intervall [0.001, 0.5] mit 3000 Gitterpunkten
x = np.linspace(0.001, 0.5, 3000) 
# Plot der Funktion auf dem Gitter
ax.plot(x, x*np.sin(1/x), color="k") 

plt.show()
```

**Mathematische Begründung:**
* Es gilt $|\sin(y)|\leq 1$ für $y\in\mathbb R$, also ist auch $f$ beschränkt: $|\sin(1/x)|\leq 1$ für alle $x\in\mathbb R_+$.
* Daher folgt $\displaystyle \lim_{x\to 0^+} x\sin(1/x)=0$.

#### e) Funktion $f:\mathbb R_+\times \mathbb R\to\mathbb R, \quad (x,y)\mapsto x^y = \exp(y\ln(x))$
* Die Funktion $f$ ist als Komposition stetiger Funktionen stetig. 
* Der Ursprung $(0,0)$ ist Häufungspunkt der Definitionsmenge $\mathbb R_+\times\mathbb R$. 
* Lässt sich die Funktion in $(x,y)=(0,0)$ stetig fortsetzen? Nein (siehe unten)
  
**Bemerkung:** Somit sollte $0^0$ nicht festgelegt werden. Man nennt $0^0$ einen **unbestimmten Ausdruck**. 

**Der Ursprung ist Häufungspunkt:** Es gilt $(t,0)\to (0,0)$ und $(\exp(-1/t),t)\to (0,0)$ und auch $(\exp(-1/t^2),t)\to (0,0)$ für $t\to 0^+$.

```python
fig, ax = plt.subplots()

# Parameter für die Darstellung der Achsen
ax.tick_params('x',labelsize=15)
ax.tick_params('y',labelsize=15)
ax.set_xlabel(r"$x$", fontsize=20) 
ax.set_ylabel(r"$y$", fontsize=20)

# Gitter auf dem Intervall [0.001, 2] mit 1000 Gitterpunkten
t = np.linspace(0.001, 2, 1000) 
# Plot der Kurven (x(t),y(t)) auf dem Gitter
ax.plot(t, 0*t)
ax.plot(np.exp(-1/t), t)
ax.plot(np.exp(-1/t**2), t)

plt.show()
```

**Mathematische Begründung:**
* Es gilt $\displaystyle\lim_{t\to 0^+} f(t,0)=1$, da für alle $t\in\mathbb R_+$ gilt $f(t,0)=1$.
* Es gilt $\displaystyle\lim_{t\to 0^+} f(\exp(-1/t),t)=\mathrm e^{-1}$, da für alle $t\in\mathbb R_+$ gilt $f(\exp(-1/t),t)=\mathrm e^{-1}$.
* Es gilt $\displaystyle\lim_{t\to 0^+} f(\exp(-1/t^2),t)=0$, da für $t\in\mathbb R_+$ gilt $f(\exp(-1/t^2),t)=\exp(-1/t)$.

```python
from matplotlib import cm

fig, ax = plt.subplots(subplot_kw={"projection": "3d"})

# Make data.
X = np.logspace(-100, 0.001, 1000)
Y = np.logspace(-100, 0.001, 1000)
X, Y = np.meshgrid(X, Y)
f = np.exp(Y*np.log(X))

# Customize the axis.
ax.set_xlabel(r"$x$", fontsize=20)
ax.set_ylabel(r"$y$", fontsize=20)
ax.set_zlabel(r"$f(x,y)$", fontsize=20)

ax.view_init(10, 220, 0)

# Plot the surface.
surf = ax.plot_surface(X, Y, f, cmap=cm.coolwarm)

plt.show()
```

Der Plot legt nahe, dass im Grenzfall jeder Wert in $[0,1]$ angenommen werden kann.

**Mathematische Begründung:** Sei $c\in\mathbb R_+$ beliebig. Dann gilt für $t\to 0^+$ 
* $(\exp(-c/t,t)\to (0,0)$ und
* $f(\exp(-c/t,t) = \exp(-c)\in (0,1)$.
* Die Grenzwerte $0$ und $1$ haben wir ja oben schon erhalten.
