# Beispiele für *sympy*
*sympy* ist eine Python-Bibliothek für symbolisches Rechnen.

Wie immer wird *sympy* zuerst als Paket eingebunden.

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

## Mathematische Ausdrücke
Im nächsten Schritt definieren wir symbolische mathematische Ausdrücke. Dazu müssen wir Symbole (*symbols*) importieren.

```python
from sympy import symbols
x, y = symbols('x y')
expr = sp.sin(x) + y
print(expr)
```

```python
expr-y
```

## Mathematische Operationen
Jetzt möchten wir Funktionen ableiten und integrieren (zuerst ohne Grenzen, dann mit). 

```python
t = sp.symbols('t')
sp.diff(sp.sin(t)*sp.exp(-t),t)
```

```python
sp.integrate(-sp.exp(-t)*sp.sin(t)+sp.exp(-t)*sp.cos(t),t)
```

```python
sp.integrate(-sp.exp(-t)*sp.sin(t)+sp.exp(-t)*sp.cos(t),(t,0,1))
```
