## reelle Folgen

```python
# Sequence x_n = 1/n
def sequence_1dn(N):
    return [1/n for n in range(1, N+1)]
```

```python
# Sequence x_n = x^n
def sequence_xpn(x, N):
    return [x**n for n in range(1, N+1)]
```

```python
# Sequence x_n = sum_1^n 1/k
def series_harmonic(N):
    x_n = [1]
    for n in range(2, N+1):
        x_n.append(x_n[-1]+1/n)
    return x_n
```

```python
# Sequence x_n = sum_1^n 1/k^2
def series_1dk2(N):
    x_n = [1]
    for n in range(2, N+1):
        x_n.append(x_n[-1]+1/n**2)
    return x_n
```

```python
# Sequence x_n = x^n for even n and 0 for uneven n
def sequence_xpn_0(x, N):
    return [x**n if n%2==0 else 0 for n in range(1, N+1) ]
```

## komplexe Folgen

```python
# Sequence z_n = z^n
def sequence_zpn(r, phi, N):
    z = complex(r*np.cos(phi),r*np.sin(phi))   
    return [z**n for n in range(1, N+1)]
```

```python
### Sequence z_n = z^n
#z_n = sequence_zpn(1,    0.2, 300)
#z_n = sequence_zpn(0.99, 0.2, 300)
#z_n = sequence_zpn(1.01, 0.2, 200)

#z_n = sequence_zpn(1,    2, 300)
#z_n = sequence_zpn(0.99, 2, 300)
z_n = sequence_zpn(1.01, 2, 300)
#z_n = sequence_zpn(1., 2*np.pi/8, 50)
```

```python
# Visualizing 
fig, ax = plt.subplots()
ax.set_aspect('equal')

x_max = max([z.real for z in z_n])
y_max = max([z.imag for z in z_n])
x_min = min([z.real for z in z_n])
y_min = min([z.imag for z in z_n])

ax.set_xlim([x_min-1,x_max+1])
ax.set_ylim([y_min-1,y_max+1])

t = np.linspace(0,2*np.pi, 1000)
ax.plot(np.cos(t), np.sin(t), "k:")

plt.show()    
for n, z in enumerate(z_n):
    if n == 0:
        ax.plot(z.real, z.imag,'r*')
    else:    
        ax.plot(z.real, z.imag,'k*')
    plt.pause(1/(n+1)**0.7)
```

## konvergente Folgen

```python
### Sequence x_n = x^n
x_n = sequence_xpn(-0.97, 100)
```

```python
# Visualizing 
fig, ax = plt.subplots()
ax.set_xlim([0, len(x_n) + 1])
d = max(x_n) - min(x_n)
ax.set_ylim([min(x_n) - d/10, max(x_n) + d/10])
ax.plot([0, len(x_n)], [0, 0], "k--")
plt.show()    
for n, x in enumerate(x_n):
    if n == 10:
        ax.plot([0, len(x_n)], [0.5, 0.5], "r:")
        ax.plot([0, len(x_n)], [-0.5, -0.5], "r:")        
    if n == 30:
        ax.plot([0, len(x_n)], [0.5, 0.5], "white")
        ax.plot([0, len(x_n)], [-0.5, -0.5], "white")
        ax.plot([0, len(x_n)], [0.3, 0.3], "r:")
        ax.plot([0, len(x_n)], [-0.3, -0.3], "r:")
    if n == 60:
        ax.plot([0, len(x_n)], [0.3, 0.3], "white")
        ax.plot([0, len(x_n)], [-0.3, -0.3], "white")
        ax.plot([0, len(x_n)], [0.1, 0.1], "r:")
        ax.plot([0, len(x_n)], [-0.1, -0.1], "r:")        
    ax.plot(n+1, x, 'k*')
    plt.pause(1 / (n+1)**0.6)
```
