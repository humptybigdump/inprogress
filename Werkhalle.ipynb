# Beispiel Werkhalle

### Definition der Verbotszonen  

Funktionale Beschreibung der ersten Verbotszone $M_0^1$

```python
g11 = lambda x: -x[0]+2
g12 = lambda x: x[0]-3
g13 = lambda x: -x[1]+1
g14 = lambda x: x[1]-5
g1 = [g11,g12,g13,g14]
g10 = lambda x: max(gi(x) for gi in g1)
```

Funktionale Beschreibung der zweiten Verbotszone $M_0^2$

```python
g21 = lambda x: -x[1]+4
g22 = lambda x: -2*x[0]+x[1]+6
g23 = lambda x: x[0]+x[1]-12
g2 = [g21,g22,g23]
g20 = lambda x: max(gi(x) for gi in g2)
```

### Plotten der Verbotszonen 

Die Verbotszonen sind durch die Mengen $(g^1_0)_\leq^0$ und $(g^2_0)_\leq^0$ definiert und können daher mit den Höhenlinien $(g^1_0)_=^0$ und $(g^2_0)_=^0$ visualisiert werden.

Berechnen der Werte von $g^1_0$ und $g^2_0$ auf $[0,9]\times[0,7]$:

```python
import numpy as np

precision = 25
k = 9*precision
l = 7*precision
x = np.linspace(0,9,k)
y = np.linspace(0,7,l)
X,Y = np.meshgrid(x,y,indexing='ij')

def eval_on_grid(g):
    Z = np.zeros((k,l))
    for i in range(k):
        for j in range(l):
            Z[i,j]=g(np.array([X[i,j],Y[i,j]]))
    return Z

Z1 = eval_on_grid(g10)
Z2 = eval_on_grid(g20)
```

Plotten der Höhenlinien und Erstellung der Abbildung

```python
from matplotlib import pyplot as plt

def plot_Werkhalle():
    for Z in [Z1,Z2]:
        plt.contour(X,Y,Z,[0], colors='black')
    plt.scatter(np.array([1,7]),np.array([1,6]), color='red')
    plt.annotate('$M_0^1$', xy=(105, 130), xycoords='figure points', fontsize=12)
    plt.annotate('$M_0^2$', xy=(255, 160), xycoords='figure points', fontsize=12)
    plt.annotate('$A$', xy=(45, 45), xycoords='figure points', fontsize=12)
    plt.annotate('$B$', xy=(270, 215), xycoords='figure points', fontsize=12)

plot_Werkhalle()
plt.grid()
plt.title('Werkhalle')
#plt.savefig('Werkhalle.png', bbox_inches="tight", dpi=600)
plt.show()
```
