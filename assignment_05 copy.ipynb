Use this behavior to calculate the population densities for the German states.

### Exercise 3: Decision Tree

Consider the following decision tree which is applied in a production's quality control:

<img align="center" width=1200 src="figures/decision_tree.png"/>

Use `if`-conditions to implement the logic of the above decision tree within a function `decision_tree()`.

The function's arguments are the attributes `material_hardness_bhn` (material hardness in Brinell) and `surface_roughness_ra_um` (surface roughness in micrometers) of a component.
The return value is the result of the quality control:
* `True`, if the quality control is passed
* `False`, if the quality control is failed

```python
def decision_tree(material_hardness_bhn , surface_roughness_ra_um):
    ### BEGIN SOLUTION
    if material_hardness_bhn <= 0.8575 and surface_roughness_ra_um <= 0.5392 and material_hardness_bhn <= -1.6170:
        return False
    if material_hardness_bhn <= 0.8575 and surface_roughness_ra_um <= 0.5392 and material_hardness_bhn > -1.6170:
        return True
    if material_hardness_bhn <= 0.8575 and surface_roughness_ra_um > 0.5392:
        return False
    if material_hardness_bhn > 0.8575:
        return True
    ### END SOLUTION
```

```python
assert decision_tree(-2, 0.5) == False
assert decision_tree(-2, 0.6) == False
assert decision_tree(0.8, 0.5) == True
assert decision_tree(0.8, 0.6) == False
assert decision_tree(2, 0.5) == True
assert decision_tree(2, 0.6) == True
```
