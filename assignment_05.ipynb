Auf solchen `Series`-Instanzen gleicher Dimension können einfache Rechenoperationen mit Hilfe der gewohnten Oparatoren (`+`, `-`, `*`, `/`) durchgeführt werden.
Nutzen Sie dieses Verhalten, um die Bevölkerungsdichte der Bundesländer effizient zu berechnen.
Speichern Sie das Ergebnis (Typ `Series`) in der Variablen `population_density`.

`<dataframe>[<spaltenname>] = <werte>`

### Aufgabe 3: Entscheidungsbaum

Betrachten Sie folgenden Entscheidungsbaum, welcher der Qualitätskontrolle einer Fertigung dient:

<img align="center" width=1200 src="figures/decision_tree.png"/>

Verwenden Sie eine Reihe von Bedingungen, um die Logik des abgebildeten Entscheidungsbaums in der Funktion `decision_tree` zu implementieren.

Argumente der Funktion sind dabei die Attribute `material_hardness_bhn` (Härte des Materials in Brinellhärte) und `surface_roughness_ra_um` (Rauigkeit der Oberfläche in Mikrometern) einer Komponente.
Rückgabe der Funktion ist das Ergebnis der Qualitätskontrolle:
* `True`, falls die Qualitätskontrolle bestanden ist ("Pass QC")
* `False`, falls die Qualitätskontrolle nicht bestanden ist ("Fail QC")

```python
def decision_tree(material_hardness_bhn , surface_roughness_ra_um):
    ### BEGIN SOLUTION
    if material_hardness_bhn <= 0.8575 and surface_roughness_ra_um <= 0.5392 and material_hardness_bhn <= -1.6170:
        return False
    if material_hardness_bhn <= 0.8575 and surface_roughness_ra_um <= 0.5392 and material_hardness_bhn > -1.6170:
        return True
    if material_hardness_bhn <= 0.8575 and surface_roughness_ra_um > 0.5392:
        return False
    if material_hardness_bhn > 0.8575:
        return True
    ### END SOLUTION
```

```python
assert decision_tree(-2, 0.5) == False
assert decision_tree(-2, 0.6) == False
assert decision_tree(0.8, 0.5) == True
assert decision_tree(0.8, 0.6) == False
assert decision_tree(2, 0.5) == True
assert decision_tree(2, 0.6) == True
```
