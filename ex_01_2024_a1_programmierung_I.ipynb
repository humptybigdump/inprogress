### Aufgabe 3: Gemischtes (4 Punkte)

Schreiben Sie in der folgenden Notebook-Zelle Python-Code der, für verschiedene Kombinationen von Seitenlängen `a` und `b` eines rechtwinkligen Dreiecks, die Länge der Hypotenuse `c` berechnet und in einem Dictionary speichert.

**a)** Dazu soll zunächst eine Funktion `pythagoras` implementiert werden, welche die Längen der Katheten eines rechtwinkligen Dreiecks `a` und `b` als Argumente entgegennimmt und die zugehörige Hypotenusen-Länge berechnet. Diese muss mit `return` zurückgegeben werden.

**b)** Die Möglichen Seitenlängen der beiden Katheten `a` und `b` des rechtwinkligen Dreiecks seien aus der Menge `{1, 2, 3, 4}`.
Für jede mögliche Kombination von Seitenlängen `a` und `b`, berechnen Sie mit Hilfe des Satzes des Pythagoras die Seitenlänge der zugehörigen Hypotenuse `c`.
Speichern Sie alle zugehörigen Ergebnisse in einem einzigen geschachteteln Dictionary `hypothenuses`.
Mit `hypothenuses[a][b]` soll anschließend die Länge der Hypotenuse eines Dreiecks mit Seiten `a` und `b` abrufbar sein.

Der letzte bereits vorhandene Code-Block gibt die Einträge von `hypothenuses` formattiert auf dem Bildschirm aus, so dass jede Zeile eines der möglichen Dreiecke enthält.
Dieser sollte ohne Anpassungen funktionieren und für die 10 möglichen Kombinationen von Seitenlängen aus `sides` die zugehörigen Dreiecksmaße ausgeben.

**Hinweise**: 
* Der Satz des Pythagoras zur Berechnung der Länge der Hypotenuse `c` lautet: $c = \sqrt{a^2 + b^2}$
* Unter der Annahme, dass `a` und `b` nicht weiter unterscheidbar sind, berechnen Sie jede mögliche Permutation des Dreiecks nur ein einziges Mal. (Ist a=1, b=2 bereits berechnet, so muss a=2, b=1 nicht erneut betrachtet werden. Somit müssen insgesamt 10 Werte berechnet werden.)

```python
from math import sqrt

def pythagoras(a, b):
# A: Implementieren Sie hier die Funktion pythagoras
### BEGIN SOLUTION
    c_squared = a**2 + b**2
    c = sqrt(c_squared)
    return c
### END SOLUTION

sides = range(1, 5, 1)

# B: Berechnen Sie hier die Hypothenusen-Längen wie in der Aufgabenstellung beschrieben
### BEGIN SOLUTION
hypothenuses = {}
for a in sides:
    hypothenuses[a] = {}
    for b in sides:
        if b <= a:
            hyp = pythagoras(a, b)
            hypothenuses[a][b] = hyp
### END SOLUTION

# Ausgabe der Ergebnisse
print(f'{"a":>10s}{"b":>10s}{"c":>10s}')
print("-"*30)

for a in hypothenuses:
    for b in hypothenuses[a]:
        print(f"{a:10d}{b:10d}{hypothenuses[a][b]:10.3f}")
```

```python
assert pythagoras
### BEGIN HIDDEN TESTS
assert abs(pythagoras(1, 2) - 2.23606797749979) < 1e-6
assert abs(pythagoras(0, 0) - 0) < 1e-6
assert abs(pythagoras(12, 12) - 16.97056274847714) < 1e-6
### END HIDDEN TESTS
```

```python
assert type(hypothenuses) == dict
for side_length in sides:
    assert side_length in hypothenuses.keys()
    assert type(hypothenuses[side_length]) == dict
### BEGIN HIDDEN TESTS
assert len(hypothenuses.keys()) == 4

for side_length in sides:
    for other_side_length in sides:
        assert other_side_length in hypothenuses[side_length].keys() or side_length in hypothenuses[other_side_length].keys()

number_of_entries = 0
for key, item in hypothenuses.items():
    number_of_entries += len(item)
assert number_of_entries >= 10

assert abs(hypothenuses[1][1] - 1.4142135623730951) < 1e-6
assert abs(hypothenuses[2][2] - 2.8284271247461903) < 1e-6
assert abs(hypothenuses[3][3] - 4.242640687119285) < 1e-6
assert abs(hypothenuses[4][4] - 5.656854249492381) < 1e-6
### END HIDDEN TESTS
```
