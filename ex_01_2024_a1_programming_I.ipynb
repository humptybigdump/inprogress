### Aufgabe 3: Mixed Topics (4 Punkte)


Write Python code in the following notebook cell that, for different combinations of side lengths `a` and `b` of a right triangle, calculates the length of the hypotenuse `c` and stores it in a dictionary.

**a)** First, implement a function `pythagoras` that takes the lengths of the two legs of a right triangle, `a` and `b`, as arguments and computes the corresponding hypotenuse length. This value should be returned using `return`.

**b)** The possible side lengths of the two legs `a` and `b` of the right triangle are from the set `{1, 2, 3, 4}`.
For each possible combination of side lengths `a` and `b`, use the Pythagorean theorem to calculate the corresponding hypotenuse length `c`.
Store all resulting values in a single nested dictionary called `hypothenuses`.
Afterward, the length of the hypotenuse for a triangle with sides `a` and `b` should be accessible using `hypothenuses[a][b]`.

The final code block already provided prints the entries of `hypothenuses` in a formatted way, so that each line shows one of the possible triangles.
This code should work without modification and output the triangle dimensions for the 10 possible combinations of side lengths from `sides`.

**Hinweise**: 
* The Pythagorean theorem for calculating the hypotenuse length `c` is: $c = \sqrt{a^2 + b^2}$
* Assuming that `a` and `b` are not distinguishable (i.e., the order doesn’t matter), you only need to compute each triangle permutation once.
(If `a=1` and `b=2` have already been computed, `a=2` and `b=1` do not need to be considered again. Thus, a total of 10 values must be computed.)

```python
from math import sqrt

def pythagoras(a, b):
# A: Implement the function pythagoras
### BEGIN SOLUTION
    c_squared = a**2 + b**2
    c = sqrt(c_squared)
    return c
### END SOLUTION

sides = range(1, 5, 1)

# B: Calculate the lengths of the hypothenuses as described
### BEGIN SOLUTION
hypothenuses = {}
for a in sides:
    hypothenuses[a] = {}
    for b in sides:
        if b <= a:
            hyp = pythagoras(a, b)
            hypothenuses[a][b] = hyp
### END SOLUTION

# Print the results
print(f'{"a":>10s}{"b":>10s}{"c":>10s}')
print("-"*30)

for a in hypothenuses:
    for b in hypothenuses[a]:
        print(f"{a:10d}{b:10d}{hypothenuses[a][b]:10.3f}")
```

```python
assert pythagoras
### BEGIN HIDDEN TESTS
assert abs(pythagoras(1, 2) - 2.23606797749979) < 1e-6
assert abs(pythagoras(0, 0) - 0) < 1e-6
assert abs(pythagoras(12, 12) - 16.97056274847714) < 1e-6
### END HIDDEN TESTS
```

```python
assert type(hypothenuses) == dict
for side_length in sides:
    assert side_length in hypothenuses.keys()
    assert type(hypothenuses[side_length]) == dict
### BEGIN HIDDEN TESTS
assert len(hypothenuses.keys()) == 4

for side_length in sides:
    for other_side_length in sides:
        assert other_side_length in hypothenuses[side_length].keys() or side_length in hypothenuses[other_side_length].keys()

number_of_entries = 0
for key, item in hypothenuses.items():
    number_of_entries += len(item)
assert number_of_entries >= 10

assert abs(hypothenuses[1][1] - 1.4142135623730951) < 1e-6
assert abs(hypothenuses[2][2] - 2.8284271247461903) < 1e-6
assert abs(hypothenuses[3][3] - 4.242640687119285) < 1e-6
assert abs(hypothenuses[4][4] - 5.656854249492381) < 1e-6
### END HIDDEN TESTS
```
