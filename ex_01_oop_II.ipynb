### Aufgabe 2: Klassen & Objekte I

**a)**

Implementieren Sie eine Klasse `Kontakt` mit den Attributen `name`, `first_name`, and `phone`.
Alle diese Attribute sollten im Konstruktor der Klasse übergeben werden können.

Implementieren Sie außerdem die Vergleichsfunktion (`__eq__`) für zwei Instanzen der Klasse, die prüft, ob die Inhalte der drei Attribute der Klasse genau übereinstimmen.

**b)**

Implementieren Sie weiterhin die Funktionen `__repr__` und `__str__` der Klasse Contact.

Die String-Repräsentation, die von `__repr__` zurückgegeben wird, sollte ein String sein, der `name`, `first_name`, and `phone` durch ein Komma getrennt enthält. Z.B., „Mustermann, Max, 12345234“

Die Methode `__str__` sollte nur den vollständigen Namen (bestehend aus den Attributen `name` und `first_name` ) der zur Instanz gehörenden Person zurückgeben. Beispiel: `Mayer, Bernd`.

```python
### BEGIN SOLUTION
class Contact:
    
    def __init__(self, name, first_name, phone):
        self.name = name
        self.first_name = first_name
        self.phone = phone
        
    def __eq__(self, other):
        if self.name != other.name:
            return False
        if self.first_name != other.first_name:
            return False
        if self.phone != other.phone:
            return False
        return True
    
    def __str__(self):
        return self.name + ", " + self.first_name
    
    def __repr__(self):
        return ", ".join([self.name, self.first_name, self.phone])
        
### END SOLUTION
```

```python
# Tests for part a)
assert Contact
assert type(Contact("Mustermann", "Max", "0721 12345678")) == Contact
assert type(Contact("Mustermann", "Max", "0721 12345678")) == Contact
contact_arne = Contact("A.", "Arne", "0721 12345678")
contact_anne = Contact("A.", "Anne", "0721 12345678")
contact_anne_2 = Contact("A.", "Anne", "0721 12345678")
assert contact_arne != contact_anne
assert contact_anne == contact_anne_2

### BEGIN HIDDEN TESTS
assert Contact("", "Anne", "") == Contact("", "Anne", "")
assert Contact("", "Anne", "a") != Contact("", "Anne", "")
assert Contact("a", "Anne", "") != Contact("", "Anne", "")
### END HIDDEN TESTS
```

```python
# Tests for part b)
contact_arne = Contact("A.", "Arne", "1234")
assert contact_arne.__repr__()
assert contact_arne.__str__()
assert contact_arne.__str__() != contact_arne.__repr__()

assert contact_arne.first_name in str(contact_arne)
assert contact_arne.name in str(contact_arne)

### BEGIN HIDDEN TESTS
contact_arne_2 = Contact("A.", "Arne", "12345")
assert str(contact_arne) == str(contact_arne_2)
assert contact_arne.__repr__() != contact_arne_2.__repr__()
assert Contact("A.", "Arne", "1234").__repr__() == contact_arne.__repr__()
### END HIDDEN TESTS
```

### Aufgabe 3: Klassen & Objekte II

Gegeben ist eine Implementierung der Klasse `Point`, die einen Punkt im dreidimensionalen Raum darstellt.

**a)**

Implementiere die Methode `euclidean_norm`, die den euklidischen Abstand vom Ursprung im dreidimensionalen Raum berechnet. Verwenden Sie diese Gleichung ist $||(x,y,z)|| = \sqrt{ x^2 + y^2 + z^2 }$



**b)**

Implementiere nun die magische Methode `__eq__`, die eine zweite Punktinstanz nimmt und prüft, ob sie gleich sind. Zwei Punkte sind identisch, wenn alle drei Koordinaten (`x`, `y` und `z`) identisch sind.

Hinweis: Verwenden Sie die Funktion `isclose` aus dem Modul `math`, um zu prüfen, ob zwei `float`-Werte ausreichend nahe beieinander liegen (z.B. `isclose(x, y, abs_tol=1e-6)`). Die erforderliche Genauigkeit wird durch das Argument `abs_tol` bestimmt und kann auf `1e-6=0.000001` gesetzt werden.


**c)**

Schließlich implementiere die Methoden `__lt__` und `__gt__`, die prüfen, ob eine Instanz kleiner bzw. größer als eine andere Instanz ist. Die Regel hier ist einfach: wenn der euklidische Abstand einer Instanz kleiner ist, dann ist die Instanz kleiner. Wenn zwei Instanzen (Punkte) den gleichen euklidischen Abstand haben, wird die Instanz mit der kleineren x-Koordinate (Attribut `x`) als kleiner angesehen. Im Falle der Gleichheit ist wiederum die Instanz mit der kleineren y-Koordinate (Attribut `y`) die kleinere.

```python
from math import isclose

class Point:
    def __init__(self, x, y, z):
        self.x = float (x)
        self.y = float (y)
        self.z = float (z)
        
    ### BEGIN SOLUTION
    def euclidean_norm(self):
        norm_sq = self.x**2 + self.y**2 + self.z**2
        return norm_sq**0.5

    def __eq__(self, other):
        return isclose(self.x, other.x, abs_tol=1e-6) and isclose(self.y, other.y, abs_tol=1e-6) and isclose(self.z, other.z, abs_tol=1e-6)
    
    def __lt__(self, other):
        norm_self = self.euclidean_norm()
        norm_other = other.euclidean_norm()
        if norm_self < norm_other:
            return True
        elif norm_self > norm_other:
            return False
        else:
            if self.x < other.x:
                return True
            elif self.x > other.x:
                return False
            else:
                return self.y < other.y
    
    def __gt__(self, other):
        return other < self and other != self
    ### END SOLUTION
```

```python
import math 
p1 = Point(1, 2, 3)
p2 = Point(1, 2, 3)

# Tests for part a)
### BEGIN HIDDEN TESTS
assert math.isclose(p1.euclidean_norm() , 3.7416573867739413, abs_tol=1e-6)
### END HIDDEN TESTS
```
