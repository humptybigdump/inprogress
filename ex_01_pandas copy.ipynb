## Which data can be processed with Pandas?

Tabular data, meaning data that comes from a spreadsheet (e.g., an Excel file) or a database (e.g., an SQL database), can be read in, analyzed, and further processed with Pandas.
The central data structure in Pandas is the ``DataFrame``.

<img align="center" src="figures/df.png" width="800" >

Important properties of DataFrames:
* A DataFrame is 2-dimensional, such that it has rows and columns and can contain different data types such as int, float, or strings.
* The DataFrame shown here has 5 columns (``matriculation_number``, ``surname``, ``course_of_study``, ``exam_grade``, ``pcc``) and 6 rows (0, ..., 4). The column names are also called column indices, and the row names are also called row indices.
* All data in a column is of the same data type or is converted to the same data type when read in. In the example, the exam grade is of type float while the name is of type string.

### Delete unnecessary rows or columns

We can use the **drop()** method to remove columns or rows from our DataFrame. 
This method takes a list of row/column names together with an ``axis`` parameter.

* The parameter ``axis=0`` or ``axis=‘index’`` indicates that we are deleting ***rows***.
* The parameter ``axis=1`` or ``axis=‘columns’`` indicates that we are deleting ***columns***.

```python
students = students.drop([4], axis = "index")
students

```

```python
students = students.drop(["pcc"], axis = "columns")
display(students)
```

### Aufgabe 4: Filtering rows

<img align="center" width=600 src="figures/data_frame_filter_row.png"/>

It is also interesting to only look at certain rows of a DataFrame.
Perhaps you only want to look at those students whose grade is better than or equal to 1.3
You can achieve this via filtering by describing the condition within the square bracket ``[]`` as follows:

```python
best_students = students[students["exam_grade"] >= 1.3]
best_students
```

To implement multiple conditions, you can implement the conditions inside parentheses and link them to each other using logical operators.

IMPORTANT: To concatenate conditions in pandas, use the **&** character instead of the **and** operator, and the **|** character (vertical bar) instead of the **or** operator.

```python
best_me_students = students[
    (students["exam_grade"] >= 1.3) & 
    (students["course_of_study"] == "ME")
]

best_me_students
```

```python
students[
    (students["exam_grade"] >= 1.3) | 
    (students["course_of_study"] == "MACH")
]
```
