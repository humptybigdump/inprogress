## Welche Daten kann man mit Pandas verarbeiten?

Tabellarische Daten, d.h. Daten, die aus einem Spreadsheet (z.B. einer Exceldatei) oder einer Datenbank (z.B. einer SQL Datenbank) kommen, können mit Pandas eingelesen, analysiert und weiterverarbeitet werden.
Die zentrale Datenstruktur in Pandas ist der ``DataFrame``.

<img align="center" src="figures/df.png" width="800" >

Wichtige Eigenschaften von DataFrames:
* Ein DataFrame ist 2-dimensional, hat also Zeilen und Spalten und kann unterschiedliche Datentypen wie int, float oder strings enthalten.
* Der hier dargestellte DataFrame hat 5 Spalten (``matriculation_number``, ``surname``, ``course_of_study``, ``exam_grade``, ``pcc``) und 6 Zeilen. Die Spaltennamen nennt man auch Spaltenindex und die Zeilennamen auch Zeilenindex.
* Alle Daten einer Spalte sind vom gleichen Datentyp oder werden beim Einlesen in den gleichen Datentyp umgewandelt. Im Beispiel ist die Klausurnote vom Typ float während der Name vom Typ String ist.

### Unnötige Zeilen oder Spalten Löschen

Mit der Methode **drop()** können wir Spalten oder Zeilen aus unserem DataFrame entfernen. 
Diese Methode nimmt eine Liste von Zeilen-/Spaltennamen zusammen mit einem ``axis``-Parameter entgegen. 

* Der Parameter ``axis=0`` oder ``axis='index'`` zeigt an, dass wir ***Zeilen*** löschen.
* Der Parameter ``axis=1`` oder ``axis='columns'`` zeigt an, dass wir ***Spalten*** löschen.

```python
students = students.drop([4], axis = "index")
students
```

```python
students = students.drop(["pcc"], axis = "columns")
display(students)
```

### Teil 4: Zeilen Filtern

<img align="center" width=600 src="figures/data_frame_filter_row.png"/>

Spannend ist es auch, nur bestimmte Zeilen eines DataFrames zu betrachten.
Zum Beispiel, wenn Sie die Studierenden betrachten möchten, deren Note besser oder gleich 1.3 ist, d.h. für die gilt, dass die Spalte `klausurnote <= 1.3` ist.
Das kann über eine Filterung erreicht werden, indem die Filter-Bedingung innerhalb einer eckigen Klammer ``[]`` wie folgt beschrieben wird:

```python
best_students = students[students["exam_grade"] >= 1.3]
best_students
```

Um mehreren Bedingungen zu implementieren, kann man die Bedingungen innerhalb Klammern implementieren und sie mit einander durch logische Operatoren verknüpfen. 

WICHTIG: um Bedingung in Pandas zu verküpfen, verwenden Sie das **&**-Zeichen anstatt des **and**-Operators, und das **|**-Zeichen (Senkrechter Strich) anstatt des **or**-Operators.

```python
best_me_students = students[
    (students["exam_grade"] >= 1.3) & 
    (students["course_of_study"] == "ME")
]

best_me_students
```

```python
students[
    (students["exam_grade"] >= 1.3) | 
    (students["course_of_study"] == "MACH")
]
```
