However, a few mistakes have crept in here...
Correct all the errors to solve the task correctly. An iterative approach makes sense here: run the code – analyze the error message – fix the error.

````python
### BEGIN SOLUTION
interest_rate = 1.05

interest_correction_by_customer = {
    "private": 0.0,
    "enterprise": 0.01,
    "charitable": 0.02
}

def revenue(cap_1, cap_2, customer_type):
    interest_rate_customer = interest_rate + interest_correction_by_customer[customer_type]
    result = interest_rate_customer * (interest_rate_customer * cap_1 + cap_2)
    return result

cap_1 = 1000
cap_2 = 500

for customer_type in interest_correction_by_customer:
    return_on_invest = revenue(cap_1, cap_2, customer_type)
    print(return_on_invest)
### END SOLUTION
````
