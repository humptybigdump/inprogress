# Error Messages and Stack Traces

### Exercise 1

The goal of this task is to show that errors are not just cryptic messages highlighted in red. On the contrary, the information they contain can be extremely useful for quickly identifying and fixing the issue.

To demonstrate this, each of the following sub-tasks contains a piece of already implemented code, which should first be copied exactly as it is into the solution cell.
However, running the code will result in an error.
Take a close look at the output in each case. Read the error message—this often already contains very useful information.
The stack trace, which shows the lines executed most recently before and during the error, often makes it possible to quickly locate the problem within the code block.

#### 1 A)

```python
name = "Antonia"
year = 2025
print(name + ", " + year)
```

````python
### BEGIN SOLUTION
name = "Antonia"
year = 2025
print(name + ", " + str(year))
### END SOLUTION
````

#### 1 B)

students = ["Anja", "Anita", "Angelo", "Alex"]
for i in range(1, 5):
    print(students[i])

#### 1 C)

```python
def func(value_a, value_b):
    return value_a / value_b

for i in range(12):
    print(func(1, i))
```

````python
### BEGIN SOLUTION
def func(value_a, value_b):
    if value_b != 0:
        return value_a / value_b
    else:
        return None

for i in range(12):
    print(func(1, i))
### END SOLUTION
````
