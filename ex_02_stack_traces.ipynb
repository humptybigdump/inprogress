# Fehlermeldungen und Stack Traces

### Aufgabe 1

Ziel dieser Aufgabe ist es, zu zeigen, dass Fehler nicht bloß rot hervorstechende kryptische Ausgaben bedeuten. Im Gegenteil können die enthaltenen Informationen extrem nützlich sein, um den Fehler schnell zu identifizieren und zu beheben.

Um dies zu demonstrieren, enthält jede der folgenden Teilaufgaben ein Stück bereits implementierten Code, welcher zunächst genau so in die Lösungszelle kopiert werden soll.
Bei Ausführung des Codes kommt es jedoch zu einem Fehler.
Schau Dir die Ausgabe hierzu jeweils genau an. Lies die Fehlermeldung, diese enthält oft bereits sehr nützliche Informationen. Der Stack Trace, welcher die bei und vor Auftritt zuletzt ausgeführten Zeilen enthält, ermöglicht es oft, den Fehler innerhalb des Codeblocks schnell zu lokalisieren.

#### 1 A)

```python
name = "Antonia"
year = 2025
print(name + ", " + year)
```

````python
### BEGIN SOLUTION
name = "Antonia"
year = 2025
print(name + ", " + str(year))
### END SOLUTION
````

#### 1 B)

```python
students = ["Anja", "Anita", "Angelo", "Alex"]
for i in range(1, 5):
    print(students[i])
```

````python
### BEGIN SOLUTION
students = ["Anja", "Anita", "Angelo", "Alex"]
for i in range(0, 4):
    print(students[i])
### END SOLUTION
````

#### 1 C)

```python
def func(value_a, value_b):
    return value_a / value_b

for i in range(12):
    print(func(1, i))
```

````python
### BEGIN SOLUTION
def func(value_a, value_b):
    if value_b != 0:
        return value_a / value_b
    else:
        return None

for i in range(12):
    print(func(1, i))
### END SOLUTION
````
