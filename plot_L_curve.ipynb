```python
import numpy as np
import matplotlib.pyplot as plt
```

load the model variations

```python
modelvar12=np.average(np.square(np.loadtxt('L12_G05/loc.mod',skiprows=2)))
modelvar17=np.average(np.square(np.loadtxt('L17_G05/loc.mod',skiprows=2)))
modelvar20=np.average(np.square(np.loadtxt('L20_G05/loc.mod',skiprows=2)))
modelvar25=np.average(np.square(np.loadtxt('L25_G05/loc.mod',skiprows=2)))
modelvar30=np.average(np.square(np.loadtxt('L30_G05/loc.mod',skiprows=2)))
modelvar=np.sqrt(np.array([modelvar12,modelvar17,modelvar20,modelvar25,modelvar30]))
```

load the misfit variations

```python
Misfit12=np.average(np.square(np.loadtxt('L12_G05/glob.res_inv')))
Misfit17=np.average(np.square(np.loadtxt('L17_G05/glob.res_inv')))
Misfit20=np.average(np.square(np.loadtxt('L20_G05/glob.res_inv')))
Misfit25=np.average(np.square(np.loadtxt('L25_G05/glob.res_inv')))
Misfit30=np.average(np.square(np.loadtxt('L30_G05/glob.res_inv')))
Misfit=np.sqrt(np.array([Misfit12,Misfit17,Misfit20,Misfit25,Misfit30]))
```

```python
damping=np.array([1.2,1.7,2,2.5,3])
```

```python
plt.plot(modelvar,Misfit)
plt.plot(modelvar,Misfit,'o')
plt.xlabel('Model variance in percent')
plt.ylabel('RMS in seconds')
for i in np.linspace(0,4,5):
    plt.text(modelvar[int(i)],Misfit[int(i)]+0.0001,str(damping[int(i)]))
```
