## 1. Control Flow and Functions

### Explanation

### Explanation

Python provides several built-in data structures that make it easy to work with collections of data.

- **Lists** are ordered and mutable. You can add, remove, or change elements.
- **Tuples** are ordered but immutable. Once created, their contents cannot be changed.
- **Dictionaries** store key-value pairs, which allow for fast lookups.

**Comprehensions** offer a compact way to create data structures using a single line of code.


````python
# List example
fruits = ["apple", "banana", "cherry"]
fruits.append("mango")
print(fruits[0])  # apple
print(fruits[-1]) # indexing with negative numbers starts from the end

# Tuple example
person = ("Alice", 24)
print(person[1])  # 24


# Dictionary example
scores = {"Alice": 92, "Bob": 85}
scores["Charlie"] = 88
print(scores["Bob"])  # 85
print(scores["Charlie"])  # 88

# List comprehension example
even_numbers = [x for x in range(10) if x % 2 == 0]
print(even_numbers)  # [0, 2, 4, 6, 8]
````

You are given two lists: one with student names and one with their corresponding scores. Your task is to:

1. Create a dictionary from the two lists and print each student's name and score
2. Extract names of students who scored above 90 using list comprehension and print them

Hint: Look up the `zip` or `enumerate` function in Python.


````python
names = ["Anna", "Ben", "Chris", "Dana"]
scores = [88, 95, 67, 91]

# YOUR CODE HERE
student_dict = {name: score for name, score in zip(names, scores)}
for name, score in student_dict.items():
    print(f"{name}: {score}")
high_scorers = [name for name, score in student_dict.items() if score > 90]
print(f"High scorers: " + ", ".join(high_scorers))
````

### Explanation

Python also supports the Object-oriented programming (OOP) paradigm. Each object is an instance of a class, which defines its structure and behavior.

To define a class, use the `class` keyword. Initialization of object attributes is done via the `__init__` method, and `self` is used to refer to the current instance. You can also define methods, which are functions associated with an object.

Inheritance allows one class to extend another. This is useful for reusing and extending functionality.


````python
import random

class Die:
    def __init__(self, sides=6):
        self.sides = sides

    def roll(self):
        return random.randint(1, self.sides)

class WeightedDie(Die):
    def __init__(self, sides=6, weights=None):
        super().__init__(sides)
        self.weights = weights or [1] * sides

    def roll(self):
        return random.choices(range(1, self.sides + 1), weights=self.weights)[0]

# Try it out
d = Die()
print(d.roll())

wd = WeightedDie(weights=[1, 1, 1, 1, 1, 10])
print(wd.roll())
````

### <span style="color:red">[TODO]</span> Exercise: Coin and BiasedCoin (4 pts)

Create two classes:

1. `Coin` that has a method `flip()` which returns either `'Heads'` or `'Tails'` with equal probability
2. `BiasedCoin` that inherits from `Coin` and allows setting the probability of `'Heads'`
3. Simulate flipping 10,000 unbiased coins and 10,000 biased coins with heads probability 0.7 and count the number of heads and tails


````python
# YOUR CODE HERE
class Coin:
    def __init__(self):
        pass

    def flip(self):
        return 'Heads' if np.random.rand() < 0.5 else 'Tails'

class BiasedCoin(Coin):
    def __init__(self, probability_heads=0.5):
        super().__init__()
        self.probability_heads = probability_heads

    def flip(self):
        return 'Heads' if np.random.rand() < self.probability_heads else 'Tails'

unbiased_coin = Coin()
biased_coin = BiasedCoin(probability_heads=0.7)

unbiased_heads = sum([unbiased_coin.flip() == 'Heads' for _ in range(10000)])
biased_heads = sum([biased_coin.flip() == 'Heads' for _ in range(10000)])

print(f"Unbiased coin heads: {unbiased_heads}")
print(f"Biased coin heads: {biased_heads}")
````

## 4. Introduction to NumPy

### Explanation

NumPy is a core library for numerical computations in Python. It provides support for:
- Multidimensional arrays (called `ndarray`)
- Efficient operations on large datasets
- Linear algebra, random sampling, and more

NumPy is optimized and implemented in C under the hood, making array computations significantly faster than using native Python lists.

To use NumPy, we typically import it using:
```python
import numpy as np
```

````python
import numpy as np

# Create arrays
arr = np.array([1, 2, 3, 4])
print("Array:", arr)

# Array of zeros and ones
zeros = np.zeros((2, 3))
ones = np.ones((2, 3))
print("Zeros:\n", zeros)
print("Ones:\n", ones)

# Random numbers
rand = np.random.rand(3)
print("Random values:", rand)

# Batch operations on arrays
print("Array + Array:\n", arr + arr)
print("Array - Array:\n", arr - arr)
print("Array * Array:\n", arr * arr)
print("Array / Array:\n", arr / arr)
print("Array ** Array:\n", arr ** arr)

# Broadcasting
print("Array + 1:\n", arr + 1)
print("1 + Array:\n", 1 + arr)
print("Array + Array:\n", arr + arr)

# Advanced indexing
print("Array > 1:\n", arr > 1)
print("Array < 1:\n", arr < 1)
print("Array == 1:\n", arr == 1)

arr[arr > 2] = 0
print("Array > 2:\n", arr)
````

## 5. Simulating Probability with NumPy

### Explanation

You can use NumPy to simulate probability experiments, such as rolling dice or flipping coins. These simulations help understand concepts like distributions and empirical probabilities, which are foundational in machine learning and statistics.

````python
# Simulate rolling a die 10000 times
rolls = np.random.randint(1, 7, size=10000)

# Frequency of each outcome
unique, counts = np.unique(rolls, return_counts=True)
probabilities = counts / 10000

print("Outcomes:", unique)
print("Estimated Probabilities:", probabilities)
````

### NumPy vs Python List Performance

A major advantage of NumPy is speed. Let’s test how fast it performs an element-wise operation compared to native Python lists.

1. Generate a large list of 1 million numbers.
2. Multiply each number by 2 using both a Python loop and a NumPy operation.
3. Compare the execution time using `%timeit`.

````python
py_list = list(range(1_000_000))

%timeit [x * 2 for x in py_list]
````

````python
np_array = np.arange(1_000_000)

%timeit np_array * 2
````


You should see a significant performance difference in favor of NumPy, which makes it the preferred choice for numerical operations in ML tasks.
