## 2.1 Arithmetic Operators

Python can perform standard arithmetic operations:

| Operator | Meaning | Example |
|:--------|:--------|:--------|
| `+` | Addition | `3 + 2` → `5` |
| `-` | Subtraction | `5 - 2` → `3` |
| `*` | Multiplication | `4 * 3` → `12` |
| `/` | Division | `10 / 2` → `5.0` |
| `**` | Exponentiation | `2**3` → `8` |
| `//` | Floor Division | `7 // 2` → `3` |
| `%` | Modulus (remainder) | `7 % 2` → `1` |

Floor division (`//`) rounds down to the nearest integer.

## 2.3 Operator Precedence

When multiple operations appear together, Python follows a specific order of precedence:

1. Parentheses `()`
2. Exponentiation `**`
3. Multiplication `*`, Division `/`, Floor Division `//`, Modulus `%`
4. Addition `+`, Subtraction `-`

Use parentheses to make the order clear and avoid mistakes.

````python
# Predict the output and then run the code

result1 = 3 + 4 * 2
result2 = (3 + 4) * 2
result3 = 2 ** 3 ** 2
result4 = (2 ** 3) ** 2

print("result1:", result1)
print("result2:", result2)
print("result3:", result3)
print("result4:", result4)
````
