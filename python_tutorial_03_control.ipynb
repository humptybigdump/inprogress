## 3.1 Conditional Statements: if, elif, else

Conditional statements allow your program to make decisions based on conditions.

Basic structure:

```python
if condition:
    statement
elif another_condition:
    another_statement
else:
    fallback_statement
```

Indentation is important in Python. Use 4 spaces for indentation.

Conditions use comparison operators:

| Operator | Meaning |
|:---------|:--------|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |

## 3.3 Loops

Loops allow repetitive execution of code blocks.

### For Loop

Iterates over a sequence like a list or range.

```python
for variable in sequence:
    statement
```

Example using `range(start, stop, step)`:

````python
# Print numbers from 1 to 5
for i in range(1, 6):
    print(i)
````

### While Loop

Repeats as long as a condition is true.

```python
while condition:
    statement
```

````python
# Simulate filling a tank with fluid until it reaches a desired level

level = 0
target_level = 10

while level < target_level:
    level += 1
    print("Current level:", level)
````

## 3.4 Loop Control Statements

- `break`: Exit the loop early.
- `continue`: Skip the rest of the current loop iteration.
- `pass`: Do nothing (use as placeholder).

Example:

````python
# Break when pressure exceeds 150 Pa

pressures = [90, 110, 130, 160, 180]

for p in pressures:
    if p > 150:
        print("Warning: High pressure detected!")
        break
    print("Pressure is normal:", p)
````
