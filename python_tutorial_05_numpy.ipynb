## 5.1 What is NumPy?

NumPy (Numerical Python) is a powerful Python library used for numerical computations.

Main features:
- Provides support for arrays and matrices
- Offers mathematical functions to operate on arrays efficiently
- Highly optimized for performance compared to pure Python lists

We import it conventionally as:

```python

```

## 5.3 Array Indexing and Slicing

Array indexing and slicing work similarly to lists, but are more powerful.

- Indexing accesses a single element.
- Slicing accesses a subset of the array.

Example:

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr[1])    # 20
print(arr[1:4])  # [20, 30, 40]
```

## 5.5 Matrices

NumPy arrays can also represent matrices (2D arrays).

Example:

```python

B = np.array([[5, 6], [7, 8]])

C = A @ B
```
