## 7.1 What is Matplotlib?

Matplotlib is the most common Python library for creating static, animated, and interactive plots.

It provides control over every element of a figure, similar to what MATLAB offers.

Import convention:

```python
import matplotlib.pyplot as plt
```

## 7.2 Plotting a Simple Graph

The basic steps to create a plot are:
- Define data points (x and y)
- Create a figure
- Plot the data
- Show the figure

Example:

```python
x = [0, 1, 2, 3]
y = [0, 1, 4, 9]
plt.plot(x, y)
plt.show()
```

## 7.3 Adding Labels, Title, and Grid

You can add more information to a plot using:
- `plt.xlabel('x-axis label')`
- `plt.ylabel('y-axis label')`
- `plt.title('Title')`
- `plt.grid(True)`

## 7.4 Multiple Lines on the Same Plot

You can plot multiple datasets on one figure by calling `plot()` multiple times before `show()`.

Example:

```python

plt.legend()
plt.show()
```

## 7.5 What is Seaborn?

Seaborn is a Python visualization library based on Matplotlib that provides a high-level interface for creating attractive and informative graphics.

It works well with Pandas DataFrames.

Import convention:

```python
import seaborn as sns
```

````python
# Import seaborn as sns
````

## 7.6 Basic Seaborn Plot

Seaborn can create a quick plot with just one line using `sns.lineplot()` or `sns.scatterplot()`.

Example:

```python
import pandas as pd

df = pd.DataFrame({
    'time': [0, 1, 2, 3, 4],
    'displacement': [0, 2, 4, 6, 8]
})
sns.lineplot(x='time', y='displacement', data=df)
plt.show()
```
