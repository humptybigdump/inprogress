## Schleifen

Eine Programiersprache wie Python einzusetzen bietet sich insbesondere an, wenn Dinge häufig wiederholt werden. Nun wäre es ineffektiv bis unmöglich den Code dafür im Programm einfach entsprechend oft zu wiederholen.  
Hier kommen _Schleifen_ zum Einsatz, die einen Programmabschnitt immer wieder wiederholen. Python kennt dabei zwei verschiedene Schleifen:

### `for`-Schleife

Eine `for`-Schleife wird immer verwendet, wenn etwas für alle Elemente einer Liste oder bestimmt oft wiederholt werden soll.

```
for i in <Liste>:
    Block
```

````python
l = [1, 2.3, "c", 4]
for i in l:
    print(i)
    print("->", 5 * i)
print("fertig")
````

Die Elemente der Liste `l` werden nacheinander der Variable `i` zugewiesen und die folgenden Zeilen mit diesem Wert ausgeführt.

_Woher weiß Python aber nun welcher Teil wiederholt werden soll?_  
Die meisten Programmiersprachen verwenden dazu Klammern nicht so Python!
Die **Einrückungen** definieren in Python Programmblöcke, dienen hier also nicht nur der besseren Lesbarkeit sondern sind Pflicht!

Inkonsistenz bei Einrückungen wird zu Fehlerausgaben führen. Glücklicherweise machen die meisten Editoren die Einrückungen schon automatisch.

````python
a = 1
 b = 2 
````

Diese `for`-Schleife iteriert also über eine Liste.
Was mache ich aber wenn ich die Anzahl der Iterationen vorgeben möchte?  
Die Funktion `range()` generiert eine Liste mit einer vorgegebenen Zahl von Elementen und kann genau für diesen Zweck benutzt werden.

````python
for i in range(10):
    print("Hallo")
````

Etwas genauer ist `range()` ein Iterator, d.h. nur die Anleitung für die Generierung der Listenelemente.

````python
range(10)
````

````python
list(range(10))  # Wieder wird ab 0 gezählt!
````

und kann mit mehreren Argumenten aufgerufen werden  
`range(<start>, <stop>, <step>)`

````python
list(range(4, 12, 2))
````

**Wichtig**: Die obere Grenze ist nicht Teil der Liste

### `while`-Schleife

Soll ein Programmabschnitt wiederholt werden, solange eine bestimmt Bedingung erfüllt ist, kommt dieser zweite Schleifentyp zum Einsatz:

```
while <Bedingung>:
    Block
```

````python
a = 0
while a < 10:
    print(a)
    a += 2
print("Fertig: a=", a)
````

### Boolsche Algebra

Bei der `while`-Schleife musste eine Bedingung angegeben werden.
Diese Bedingungen können nur wahr oder falsch sein.

````python
print(1 < 5)
print(3 > 5)
````

Die Vergleiche `<`, `>`, `<=` und `>=` erschliessen sich sofort.
Auf Gleichheit prüft man mit `==`, auf Ungleichheit mit `!=`.

Weisst man das Ergebnis einer Variable zu, so hat diese einen Datentyp, der oben nur kurz erwähnt wurde, `bool`.

````python
a = 3 > 4
print(a)
type(a)
````

Durch logische Operatoren lassen sich boolsche Ausdrücke auch miteiner Verknüpfen. Hierbei gibt es drei Operatoren:

Das Ergebnis ist _true_ wenn gilt:

- `and`: beides ist wahr
- `or`: mindestens ein Teil ist wahr
  und
- `not`: negiert

````python
a = 10
print(a < 20 and a > 5)
print(a < 20 and a > 12)
print(a < 20 or a > 12)
print(not a == 10)
````

## Bedingte Anweisungen, Verzweigungen

Mit diesen Bedingungen kann nun auch der Programmablauf gesteuert werden. Häufig soll ein Programmteil nur ausgeführt werden, wenn eine Bedingung erfüllt ist, oder es soll im Programm verzweigt werden.  
Hierfür verwendet man die `if`-`else`-Anweisung.

```
if <Bedingung>:
    Block
elif <Bedingung>:
    Block
else:
    Block
```

````python
a = 30
if a > 20:
    print("a ist größer als 20")
else:
    print("a ist kleiner 20")
````

**Auch hier ist wieder die Einrückung entscheidend!**  
Die Kombination aus `else` und `if` kann durch `elif` kürzer und übersichtlicher formuliert werden;

````python
a = 20
if a > 20:
    print("a ist größer als 20")
elif a == 20:
    print("a ist gleich 20")
else:
    print("a ist kleiner als 20")
````

## Funktionen

Bei Programmabschnitten die immer wieder an verschiedenen Stellen verwendet werden, ist es sinnvoll diese in eigene Funktionen auszulagern. Dies spart nicht nur Tipparbeit, sondern macht das Programm auch wesentlich übersichtlicher und leichter "wartbar".

Eigene Funktionen definiert man in Python mit dem Schlüsselwort `def`:

```
def <FUNKTIONSNAME>([<ARGUMENTE]):
    <PROGRAMMBLOCK>
    return <RÜCKGABE>
```

wie immer ist die Einrückung hier wichtig.  
Funktionsdefinitionen sollten immer **vor** dem eigentlichen Programm stehen.

````python
def quadrat():
    a = 4 * 4
    return a
````

````python
quadrat()
````

````python
def quadrat(x):
    a = x * x
    return a
````

Funktionen können auf Variablen, die ausserhalb definiert wurden (_globale Variablen_) zugreifen.
**Dies ist sehr unverständlich und fehleranfällig und sollte wenn möglich vermieden werden!**

````python
def addition(x):
    foo = 10
    a = x + c  # c ist hier nicht definiert, der globale Wert wird verwendet
    return a
````

Die Funktion übernimmt den globalen Wert von `c`

````python
c = 2
addition(4)
````

Variablen, die innerhalb einer Funktion definiert werden, sind aber ausserhalb nicht bekannt

````python
print(foo)
````
