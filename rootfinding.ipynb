```python
import matplotlib.pyplot as plt
import scipy.optimize
import numpy as np

# function for which we want to find the root
def f(x, a, b):
    return a * x**3 - b * x + 10

a = 2
b = 3

x_vals = np.linspace(-3, 2, 100)
y_vals = f(x_vals, a, b)

plt.figure()
plt.plot(x_vals, y_vals, color='black')
plt.axhline(y=0, color='grey', linestyle='dashed')
plt.show()

sol = scipy.optimize.root(f, x0=np.array([-1.0]), args=(a, b))
print(sol.x)

plt.figure()
plt.plot(x_vals, y_vals, color='black')
plt.axhline(y=0, color='grey', linestyle='dashed')
plt.axvline(x=sol.x, color='red')
plt.show()
```
