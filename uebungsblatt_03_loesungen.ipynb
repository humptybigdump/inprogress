#### 1 A)

Im Folgenden betrachten wir die Klasse `Smart_Device`, die ein intelligentes Smart Home Gerät abstrahiert.

Die Klasse hat ein boolesches Attribut „on“, das angibt, ob das Gerät gerade ein- oder ausgeschaltet ist. 
Außerdem gibt es ein Attribut `max_energy_consumption_kwh`, das den maximalen Stromverbrauch des Geräts in Kilowatt angibt.

Was Sie tun müssen:
- Implementieren Sie die Methode `turn_on_off` in der Klasse, die den Wert des booleschen Attributs `on` von `False` auf `True` umkehrt und umgekehrt.
- Erstelle eine Instanz von `Smart_Device`, deren Name `my_device` ist, und `max_energy_consumption_kW` ist 30.
- Benutze dann die Methode `turn_on_off`, um es einzuschalten. Schließlich wird der Wert des Attributs `on` ausgedruckt.

```python
class Smart_Device:
    
    def __init__(self, max_energy_consumption_kW):
        self.on = False
        self.max_energy_consumption_kW = max_energy_consumption_kW
        
    ### BEGIN SOLUTION
    def turn_on_off(self):
        self.on = not self.on
    ### END SOLUTION

my_device = None

### BEGIN SOLUTION
my_device = Smart_Device(30)
my_device.turn_on_off()
print(my_device.on)
### END SOLUTION
```

#### 1 B)

In diesem Teil werden wir eine Klasse namens `Smart_Light` implementieren, die von `Smart_Device` erbt.

Was Sie tun müssen:
- Implementieren Sie die Klasse `Smart_Light`, die von der Klasse `Smart_Device` erbt
- Implementiere einen Konstruktor, der die gleichen Argumente wie die Superklasse hat, zusätzlich zu zwei neuen Attributen, nämlich `brightness` und `color`.
- Setzen Sie die Standard-Helligkeit auf 70 und die Standard-Farbe auf weiß.

Tipp: Verwenden Sie die Funktion `super()`, um die Variablen der Superklasse zu initialisieren.

```python
### BEGIN SOLUTION
class Smart_Light(Smart_Device):
    def __init__(self, max_energy_consumption_kW, brightness=70, color="white"):
        super().__init__(max_energy_consumption_kW)
        self.brightness = brightness
        self.color = color
### END SOLUTION
```

```python
import inspect 

assert Smart_Light
assert issubclass(Smart_Light, Smart_Device)

signature = inspect.signature(Smart_Light.__dict__["__init__"])
params = signature.parameters.values()

param_names = [param.name for param in params]
assert "self" in param_names
assert "max_energy_consumption_kW" in param_names
assert "brightness" in param_names
assert "color" in param_names

assert signature.parameters["brightness"].default == 70
assert signature.parameters["color"].default == "white"

print("TEST PASSED!")
```

#### 2 A)
Gegeben ist die Klasse `Robot`, die einen Roboter abstrahiert. Die Klasse hat die Attribute `name`, `battery_capacity` und `task`.

Implementieren Sie die Methoden `__str__` und `__repr__` wie folgt:
- Die magische Methode ``__str__`` soll einen String zurückgeben als `my name is <name>, my task is <task> and i can work for <battery_capacity> hours`
- Die magische Methode ``__repr__`` sollte eine String zurückgeben, die aus `name`, `task` und `battery_capacity` besteht. (z.B., `"spot,cleaning,50"`)

Erstelle nun eine Instanz mit dem Namen `my_robot`, deren Name `Jack` ist und deren Aufgabe `teaching` ist, und die eine Batteriekapazität von `12` Stunden hat. Schließlich wandeln Sie diese Klasse in eine String um und speichern sie in einer neuen Variablen namens `str_jack`. 

```python
class Robot:
    
    def __init__(self, name, battery_capacity=50, task="cleaning"):
        self.name = name
        self.battery_capacity = battery_capacity
        self.task = task
        
    def __str__(self):
        ### BEGIN SOLUTION
        return f"my name is {self.name}, my task is {self.task} and I can do work for {self.battery_capacity} hours"
        ### END SOLUTION

    def __repr__(self):
        ### BEGIN SOLUTION
        return ",".join([self.name, self.task, str(self.battery_capacity)])
        ### END SOLUTION

my_robot = None
str_jack = None
### BEGIN SOLUTION
my_robot = Robot("Jack", 12, "teaching")
str_jack = str(my_robot)
print(str_jack)
my_robot.__repr__()
### END SOLUTION
```

```python
assert my_robot
assert str_jack

assert isinstance(my_robot, Robot)
assert str_jack == "my name is Jack, my task is teaching and I can do work for 12 hours"

assert (my_robot.__repr__() == "Jack,teaching,12") or (my_robot.__repr__() == "Jack, teaching, 12")
print("TEST PASSED!")
```

#### 2 B)
Gegeben ist die Klasse `Grade`, die die Note eines Schülers abstrahiert. Die möglichen Noten sind `A+`, `B+`, `C+`, `D`, wobei `A+` die beste Note und `D` die schlechteste Note ist. 

Wir wollen die Methoden `__lt__`, `__gt__` und `__eq__` implementieren, um die Noten miteinander zu vergleichen. Zum Beispiel, wenn wir `A+` und `B+` vergleichen:
- `A+` > `B+` muss `True` zurückgeben
- `A+` < `B+` muss `False` zurückgeben
- `A+` == `B+` muss `False` zurückgeben

Hinweise:
- Sie können im Konstruktor zusätzliche Attribute initialisieren. Die Argumente müssen jedoch dieselben bleiben.
- Vielleicht können Ihnen Wörterbücher helfen?

```python
class Grade():
    possible_grades = ["A+", "B+", "C+", "D"]
    def __init__(self, grade):
        if grade in self.possible_grades:
            self.grade = grade
        else:
            raise ValueError(f"grade must be on in {self.possible_grades}")
        ### BEGIN SOLUTION
        self.numeric_grade_dict = {
            "A+": 4,
            "B+": 3,
            "C+": 2,
            "D": 1
        }
        ### END SOLUTION

    ### BEGIN SOLUTION
    def __eq__(self, other):
        return self.numeric_grade_dict[self.grade] == self.numeric_grade_dict[other.grade]

    def __lt__(self, other):
        return self.numeric_grade_dict[self.grade] < self.numeric_grade_dict[other.grade]

    def __gt__(self, other):
        return self.numeric_grade_dict[self.grade] > self.numeric_grade_dict[other.grade]
    ### END SOLUTION
```

```python
assert Grade

assert Grade("A+") == Grade("A+")
assert Grade("B+") == Grade("B+")
assert Grade("C+") == Grade("C+")
assert Grade("D") == Grade("D")

assert Grade("A+") != Grade("B+")
assert Grade("A+") != Grade("C+")
assert Grade("A+") != Grade("D")
assert Grade("B+") != Grade("C+")
assert Grade("B+") != Grade("D")
assert Grade("C+") != Grade("D")

assert Grade("A+") > Grade("B+")
assert Grade("A+") > Grade("C+")
assert Grade("A+") > Grade("D")
assert Grade("B+") > Grade("C+")
assert Grade("B+") > Grade("D")
assert Grade("C+") > Grade("D")

assert not Grade("A+") < Grade("B+") 
assert not Grade("A+") < Grade("C+")
assert not Grade("A+") < Grade("D")
assert not Grade("B+") < Grade("C+")
assert not Grade("B+") < Grade("D")
assert not Grade("C+") < Grade("D")

print("TEST PASSED!")
```
