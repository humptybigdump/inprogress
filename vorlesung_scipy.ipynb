## scipy.integrate
Viele dieser Methoden sind sehr fortgeschritten, hier
wird das Modul nur an Beispiel des *integrate*-Submoduls vorgestellt.<br>
Zuerst binden wir *numpy* und das *integrate*-Submodul ein.

```python
import numpy as np
import scipy.integrate as integrate
```

*integrate* bietet verschiedene Methoden zur numerischen Berechnungen von Integralen.<br>
Wir benötigen noch eine Funktion die integriert werden soll. Diese wird als Python-Funktion definiert.

```python
def f(x):
    return x**2
```

Jetzt wird die `quad()` Funktion des Submoduls verwendet um die Funktion zu integrieren.
`quad()` benötigt dazu die Funktion und die Integrationsgrenzen.

```python
integrate.quad(f, 0, 3)
```

`quad()` gibt zwei Werte zurück: Zuerst das Ergebnis, dann eine Abschätzung für die Ungenauigkeit des Wertes.

Diese Funktion hätten wir auch leicht im Kopf integrieren können. Wir rechnen mal kurz nach

```python
def int(x):
    return x**3 / 3
        
```

```python
int(3) - int(0)
```

`quad()` kann auch uneigentliche Integrale berechnen.
Ein typisches Beispiel ist hier das Gaussche Integral:

```python
def gauss(x):
    return np.exp(-x**2) / np.sqrt(np.pi)
```

```python
integrate.quad(gauss, -np.inf, np.inf)
```
