# Sympy

```python
import numpy as np
import sympy
```

Einen ersten Unterschied zu den numerischen Modulen sieht man wenn man bei Berechnungen, deren Ergebnis eine irrationale Zahl sind.

```python
np.sqrt(8)
```

```python
sympy.sqrt(8)
```

Da sich das Ergebnis nicht exakt als Fließkommazahl darstellen läßt, führt SymPy die Wurzel nicht aus.
Dadurch ergibt sich kein Rundungsfehler.

```python
print(np.sqrt(8)**2)
print(sympy.sqrt(8)**2)
```

Will man in einem solchen Fall den (genäherten) Zahlenwert, so muss man das explizit mit der Methode evalf() fordern.

```python
sympy.sqrt(8).evalf()
```

Der zentrale Zweck von SymPy sind allerdings analytische, symbolisch Rechnungen.<br>
Nun werden in Python allerdings Buchstaben wie `x` als Variablennamen verstanden. 
Es Bedarf also eines Tricks um damit irgendwie symbolisch zu rechnen.

Mit der SymPy-Funktion `symbols()` kann man Werte des Typs `Symbol` erzeugen und diese Variablen zuweisen.


```python
a, x = sympy.symbols('a, x')
type(a)
a
```

Die Variablen `a` und `x` enthalten jetzt die SymPy-Symbole *a* und *x*.<br>
Eine alternative Möglichkeit ist der Import der Symbole aus dem Submodul `sympy.abc`.

```python
from sympy.abc import b, c 
type(b)
type(c)
```

Zu beachten ist allerdings, dass man immer noch die Variablen wieder mit anderen Werten füllen kann.

```python
c = 1
type(c)
```

Auch verbietet einem niemand verwirrende Dinge wie das Folgende zu machen

```python
v, w = sympy.symbols('w, v') 
v
```

## Analytische Ausdrücke, differenzieren, integrieren
Hat man Symbole, kann man daraus auch mathematische Ausdrücke basteln

```python
expr = x**2 + x + a
expr
```

und diese z.B. differenzieren

```python
sympy.diff(expr, x)
```

oder integrieren

```python
sympy.integrate(expr, x)
```

Bei bestimmten Integralen gibt man einfach nur auch die Grenzen an

```python
int = sympy.integrate(expr, [x, 0, 2])
int
```

## Werte einsetzen
Oft will man bei einer Rechnung Ausdrücke für bestimmte Zahlenwerte von Variablen berechnen.
Diese Funktion bietet die Methode `subs()`.<br>
`subs()` verändert dabei die Variable selbst nicht

```python
int.subs(a, 1/3)
a
```

Wobei man schon ein kleines Problem sieht: Python berechnet `1/3` zu einer Fließkommazahl und erst dann findet die Ersetzung statt. <br>
Hier hilft:

```python
int.subs(a, '1/3')
```

IdR. muss man Brücke explizit mit der Funktion `Rational()` als solche erzwingen.

```python
sympy.Rational(2, 6)
```

## simplify
Von sich aus vereinfacht SymPy keine Ausdrücke

```python
f = sympy.sin(x)**2 + sympy.cos(x)**2 
f
```

 Achtung: Die Schreibweise $\sin^2(x)$ versteht Python nicht. Auch wenn Sympy das hier in der Darstellung verwendet

Will man Ausdrücke vereinfachen, muss man dies explizit mit der Funktion `simplify()` erzwingen

```python
sympy.simplify(f)
```

oder objectorientiert die Methode `simplify()` anwenden.

```python
f.simplify()
```
